```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import torchvision
import torchvision.transforms.functional as TF
from torchvision.utils import draw_bounding_boxes
from PIL import Image

# Import native NMS
from LCRP.models.yolov5 import non_max_suppression

# ==========================================
# 1. Helper Functions (Exact match to infer.py)
# ==========================================
def generate_positions(length, tile, stride):
    if length <= tile: return [0]
    positions, pos = [], 0
    while True:
        positions.append(pos)
        if pos + tile >= length: break
        pos += stride
        if pos + tile > length: pos = length - tile
    return positions

def tile_ownership_region(x0, y0, tile, W, H, margin):
    is_left, is_top = (x0 == 0), (y0 == 0)
    is_right, is_bottom = (x0 + tile >= W), (y0 + tile >= H)
    left_m = 0 if is_left else margin
    top_m = 0 if is_top else margin
    right_m = 0 if is_right else margin
    bottom_m = 0 if is_bottom else margin
    ox1 = max(0, min(x0 + left_m, W))
    oy1 = max(0, min(y0 + top_m, H))
    ox2 = max(0, min(x0 + tile - right_m, W))
    oy2 = max(0, min(y0 + tile - bottom_m, H))
    return ox1, oy1, ox2, oy2

def global_nms_xyxy(det, iou_thres=0.25):
    if det is None or len(det) == 0:
        return det
    boxes = det[:, :4]
    scores = det[:, 4]
    classes = det[:, 5]
    keep = torchvision.ops.batched_nms(boxes, scores, classes, iou_thres)
    return det[keep]

# ==========================================
# 2. Load the RAW Full-Resolution Image
# ==========================================
raw_dataset = PersonCarDataset(root_dir=root_dir, split="train", transform=None)
sample_idx = 0  
out = raw_dataset[sample_idx]

try:
    raw_img, _ = out
except Exception:
    raw_img, _ = out[0], out[1]

if isinstance(raw_img, torch.Tensor):
    raw_img = TF.to_pil_image(raw_img)
elif isinstance(raw_img, np.ndarray):
    raw_img = Image.fromarray(raw_img)

W, H = raw_img.size
print(f"Original Full Image Size: {W}x{H}")

# ==========================================
# 3. Setup Tiling Parameters
# ==========================================
tile_size = 640
overlap_px = 64
stride = tile_size - overlap_px
margin = max(0, overlap_px // 2)

xs = generate_positions(W, tile_size, stride)
ys = generate_positions(H, tile_size, stride)

print(f"Scanning grid: {len(ys)} rows by {len(xs)} columns ({len(ys)*len(xs)} tiles)...")

# ==========================================
# 4. Run Tiled Inference (Tile NMS first!)
# ==========================================
all_det = []
model.eval()
device = next(model.parameters()).device

with torch.no_grad():
    for y0 in ys:
        for x0 in xs:
            tile = raw_img.crop((x0, y0, x0 + tile_size, y0 + tile_size))
            tile_np = np.array(tile)
            
            # Pad edges
            th, tw = tile_np.shape[:2]
            if th != tile_size or tw != tile_size:
                padded = np.full((tile_size, tile_size, 3), 114, dtype=tile_np.dtype)
                padded[:th, :tw] = tile_np
                tile_np = padded
                
            img_tensor = TF.to_tensor(tile_np).unsqueeze(0).to(device)
            
            # 1. Raw Prediction
            raw_output = model.raw_predict(img_tensor)
            raw_pred = raw_output[0] if isinstance(raw_output, (tuple, list)) else raw_output
            
            # 2. TILE-LEVEL NMS (Matches your infer.py arg parser default conf_thres=0.35)
            det_tile = non_max_suppression(raw_pred, conf_thres=0.35, iou_thres=0.45)[0]
            
            if len(det_tile) == 0:
                continue
                
            # 3. Shift clean boxes to global image space (det_tile is x1, y1, x2, y2)
            det_tile[:, 0] += x0
            det_tile[:, 2] += x0
            det_tile[:, 1] += y0
            det_tile[:, 3] += y0
            
            # Clamp to bounds
            det_tile[:, 0].clamp_(0, W)
            det_tile[:, 1].clamp_(0, H)
            det_tile[:, 2].clamp_(0, W)
            det_tile[:, 3].clamp_(0, H)
            
            # 4. Tile Ownership filter (prevent duplicate boxes on seams)
            ox1, oy1, ox2, oy2 = tile_ownership_region(x0, y0, tile_size, W, H, margin)
            cx = 0.5 * (det_tile[:, 0] + det_tile[:, 2])
            cy = 0.5 * (det_tile[:, 1] + det_tile[:, 3])
            
            keep = (cx >= ox1) & (cx < ox2) & (cy >= oy1) & (cy < oy2)
            det_tile = det_tile[keep]
            
            if len(det_tile) > 0:
                all_det.append(det_tile)

# ==========================================
# 5. Global NMS & Plotting
# ==========================================
if len(all_det) > 0:
    det = torch.cat(all_det, dim=0)
    
    # Global NMS (Normal)
    det = global_nms_xyxy(det, iou_thres=0.45)
    
    # Global NMS (Merge seams)
    final_det = global_nms_xyxy(det, iou_thres=0.1)
else:
    final_det = torch.empty((0, 6))

print(f"Total Objects Detected After Stitching & NMS: {len(final_det)}")

img_display = TF.to_tensor(raw_img)
img_display = (img_display * 255).to(torch.uint8)

if len(final_det) > 0:
    coords = final_det[:, :4].cpu()
    result_img = draw_bounding_boxes(img_display, coords, colors="red", width=5)
else:
    result_img = img_display

plt.figure(figsize=(12, 12), dpi=200)
plt.imshow(result_img.permute(1, 2, 0).numpy())
plt.axis("off")
plt.title("YOLOv6 Tiled Inference & Stitched (Exact Replica)")
plt.tight_layout()
plt.show()
```

```python
import os
import gc
import numpy as np
import torch
import torchvision.transforms.functional as TF
import torchvision
from torchvision.utils import draw_bounding_boxes, make_grid
import matplotlib.pyplot as plt
from PIL import Image
import zennit.image as zimage
from crp.concepts import ChannelConcept
from crp.helper import get_layer_names

from LCRP.utils.crp_configs import ATTRIBUTORS, CANONIZERS, VISUALIZATIONS, COMPOSITES

# ==========================================
# HELPER FUNCTIONS 
# ==========================================
def _to_2d(array):
    arr = np.array(array)
    arr = np.squeeze(arr)
    if arr.ndim == 2:
        return arr
    if arr.ndim == 3:
        if arr.shape[0] <= 4 and arr.shape[1] > 10 and arr.shape[2] > 10:
            return arr.max(axis=0)
        if arr.shape[2] <= 4 and arr.shape[0] > 10 and arr.shape[1] > 10:
            return arr.mean(axis=2)
        if arr.shape[0] == 1:
            return arr[0]
        return arr.mean(axis=0)
    shape = arr.shape
    idx = list(np.argsort(shape)[-2:])
    perm = [i for i in range(arr.ndim) if i not in idx] + idx
    arr = np.transpose(arr, perm)
    while arr.ndim > 2:
        arr = arr.mean(axis=0)
    return arr

def _colorize_signed_map(raw_map, cmap_name="seismic", clip_percentile=99.5, power=1.15, saturation=0.85):
    map2d = _to_2d(raw_map).astype("float32")
    if map2d.size == 0:
        return np.zeros((1, 1, 3), dtype=np.float32)

    abs_values = np.abs(map2d)
    scale = np.percentile(abs_values, clip_percentile)
    if scale <= 0:
        scale = abs_values.max()
    if scale <= 0:
        return np.zeros(map2d.shape + (3,), dtype=np.float32)

    normalized = np.clip(map2d / (scale + 1e-12), -1.0, 1.0)
    normalized = np.sign(normalized) * (np.abs(normalized) ** power)

    cmap = plt.colormaps.get_cmap(cmap_name)
    colored = cmap((normalized + 1.0) / 2.0)[..., :3]
    colored = colored * saturation + (1.0 - saturation)
    return colored


# ==========================================
# MAIN CRP PIPELINE
# ==========================================
def optimized_auto_stitched_crp(model_name, model, raw_dataset, sample_id, class_id, layer, mode, n_concepts, n_refimgs, output_dir):
    device = next(model.parameters()).device
    model = model.to(device).eval()

    # 1. Load Full Image
    out = raw_dataset[sample_id]
    full_img = out[0] if isinstance(out, (list, tuple)) else out
    
    if isinstance(full_img, torch.Tensor):
        full_img_pil = TF.to_pil_image(full_img)
    elif isinstance(full_img, Image.Image):
        full_img_pil = full_img
    else:
        full_img_pil = Image.fromarray(np.array(full_img))

    W, H = full_img_pil.size
    tile_size = 640
    stride = 576
    
    # 2. FAST SCAN
    print("Scanning 4K image for targets...")
    best_x0, best_y0 = None, None
    target_box = None
    
    xs = list(range(0, W, stride))
    ys = list(range(0, H, stride))
    if xs[-1] + tile_size > W: xs[-1] = max(0, W - tile_size)
    if ys[-1] + tile_size > H: ys[-1] = max(0, H - tile_size)

    with torch.no_grad():
        for y0 in ys:
            for x0 in xs:
                tile_pil = full_img_pil.crop((x0, y0, x0 + tile_size, y0 + tile_size))
                tile_tensor = TF.to_tensor(tile_pil).unsqueeze(0).to(device)
                
                preds = model.predict_with_boxes(tile_tensor)
                if len(preds[1][0]) > 0:
                    predicted_classes = preds[0].argmax(dim=2)[0]
                    boxes = preds[1][0]
                    
                    for b, c in zip(boxes, predicted_classes):
                        if c == class_id:
                            best_x0, best_y0 = x0, y0
                            target_box = b.clone().detach().cpu()
                            break
                if best_x0 is not None:
                    break
            if best_x0 is not None:
                break

    if best_x0 is None:
        print(f"No objects of class {class_id} found in this image.")
        return

    print(f"Target found at tile X={best_x0}, Y={best_y0}. Running heavy CRP...")

    # 3. RUN CRP ONLY ON THE DETECTED TILE
    tile_pil = full_img_pil.crop((best_x0, best_y0, best_x0 + tile_size, best_y0 + tile_size))
    tile_tensor = TF.to_tensor(tile_pil).unsqueeze(0).to(device)

    composite = COMPOSITES[model_name](canonizers=[CANONIZERS[model_name]()])
    condition = [{"y": class_id}]
    attribution = ATTRIBUTORS[model_name](model)
    
    x = tile_tensor.detach().requires_grad_(True)
    attr = attribution(x, condition, composite, record_layer=[layer], init_rel=1)

    raw_heat = attr.heatmap.detach().cpu().numpy()
    tile_heatmap = _colorize_signed_map(raw_heat)

    # Shift bounding box
    target_box[0] += best_x0
    target_box[2] += best_x0
    target_box[1] += best_y0
    target_box[3] += best_y0
    box_tensor = target_box[:4].unsqueeze(0)

    # 4. CPU STITCHING (Global Projection)
    print("Stitching CRP to 4K canvas...")
    full_raw_display = np.array(full_img_pil).astype(np.float32) / 255.0

    full_heatmap_display = np.zeros_like(full_raw_display)
    th, tw = tile_heatmap.shape[:2]
    full_heatmap_display[best_y0:best_y0+th, best_x0:best_x0+tw] = tile_heatmap

    base_img_uint8 = TF.to_tensor(full_img_pil).mul(255).byte()
    result = draw_bounding_boxes(base_img_uint8, box_tensor, colors=["#ffcc00"], width=8)
    full_segmentation_display = np.array(TF.to_pil_image(result)).astype(np.float32) / 255.0

    # 5. Extract Concepts & Ref Images
    print("Extracting Concept Reference Images...")
    if mode == "relevance":
        channel_rels = ChannelConcept().attribute(attr.relevances[layer], abs_norm=True)
    else:
        channel_rels = attr.activations[layer].detach().cpu().flatten(start_dim=2).max(2)[0]
        channel_rels = channel_rels / channel_rels.abs().sum(1)[:, None]

    topk = torch.topk(channel_rels[0], n_concepts)
    topk_ind = topk.indices.detach().cpu().numpy()

    # --- BUG FIX: Create a Tensor-yielding wrapper dataset for the Reference Generator ---
    class RefDatasetWrapper(torch.utils.data.Dataset):
        def __init__(self, base_dataset):
            self.base_dataset = base_dataset
            # LCRP needs access to class_names
            self.class_names = getattr(base_dataset, 'class_names', ['person', 'car']) 
            
        def __len__(self):
            return len(self.base_dataset)
            
        def __getitem__(self, idx):
            out = self.base_dataset[idx]
            img = out[0] if isinstance(out, (list, tuple)) else out
            target = out[1] if isinstance(out, (list, tuple)) else None
            
            # Force conversion to Tensor BEFORE LCRP touches it
            if isinstance(img, torch.Tensor):
                img = TF.resize(img, (640, 640), antialias=True)
            elif isinstance(img, Image.Image):
                img = TF.to_tensor(TF.resize(img, (640, 640)))
            else:
                img = TF.to_tensor(TF.resize(Image.fromarray(np.array(img)), (640, 640)))
                
            return img, target

        # --- NEW: Safely pass through LCRP's required methods ---
        def reverse_normalization(self, x):
            if hasattr(self.base_dataset, 'reverse_normalization'):
                return self.base_dataset.reverse_normalization(x)
            return x

        def reverse_augmentation(self, x):
            if hasattr(self.base_dataset, 'reverse_augmentation'):
                return self.base_dataset.reverse_augmentation(x)
            return x
            
        def __getattr__(self, name):
            # If LCRP asks for anything else, fetch it from the original dataset
            return getattr(self.base_dataset, name)

    # Wrap the raw dataset
    tensor_ref_dataset = RefDatasetWrapper(raw_dataset)

    from src.pcx_helper import get_ref_images
    ref_imgs_save_path = os.path.join(os.path.dirname(output_dir), "ref_imgs", f"ref_imgs_{n_refimgs}")
    attribution_ref = ATTRIBUTORS[model_name](model)
    fv_ref = VISUALIZATIONS[model_name](
        attribution_ref, tensor_ref_dataset, get_layer_names(model, [torch.nn.Conv2d]),
        preprocess_fn=lambda img: img, path=output_dir, max_target="max", device=str(device)
    )
    ref_imgs = get_ref_images(fv_ref, topk_ind.tolist(), layer, composite=composite, class_id=class_id, n_ref=n_refimgs, ref_imgs_save_path=ref_imgs_save_path)

    # 6. Plot the Final 4K Image
    fig, axs = plt.subplots(1, 2, figsize=(24, 12), dpi=150)

    axs[0].set_title(f"Original 4K Image - Detection (Class {class_id})", fontsize=16)
    axs[0].imshow(full_segmentation_display)
    axs[0].axis("off")

    axs[1].set_title("Global CRP Heatmap Projection", fontsize=16)
    axs[1].imshow(full_raw_display)
    axs[1].imshow(full_heatmap_display, alpha=0.6)
    axs[1].axis("off")

    plt.tight_layout()
    plt.show()
    
    del x, attr, attribution, tile_tensor, full_img_pil, base_img_uint8
    torch.cuda.empty_cache()
    gc.collect()

# --- Run it ---
raw_dataset = PersonCarDataset(root_dir=root_dir, split="train", transform=None)

class_id = 0 
sample_id = 0 
n_concepts = 3 
n_refimgs = 12 
layer = "module.backbone.ERBlock_3.0.rbr_dense.conv" 
mode = "relevance"

optimized_auto_stitched_crp(model_name, model, raw_dataset, sample_id, class_id, layer, mode, n_concepts, n_refimgs, output_dir)
```
